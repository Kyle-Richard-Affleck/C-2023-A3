```python
import numpy as np
import os
import fnmatch
from astropy.io import fits
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
from photutils.centroids import centroid_2dg
import matplotlib.gridspec as gridspec
from scipy.ndimage import map_coordinates
from scipy.ndimage import shift
import astropy.units as u
from sbpy.activity import Afrho, Efrho
from sbpy.data import Ephem
from sbpy.calib import solar_fluxd
from scipy.ndimage import gaussian_filter1d
from scipy.signal import savgol_filter
from astropy.coordinates import SkyCoord
from astropy import units as u
from scipy.ndimage import gaussian_filter
from astropy.stats import SigmaClip
from photutils.background import Background2D, MedianBackground
from mpl_toolkits.axes_grid1 import make_axes_locatable
from scipy.signal import find_peaks
import matplotlib.colors as mcolors
# =============================== Configuration ===============================

# Change these two values only
obs_date_str = '20241101'  # Directory date
plot_date_str = '01 11 24'  # Date for the plot titles & file names

rundir = f'/Users/kylerichardaffleck/Downloads/C2023A3/{obs_date_str}'
output_dir = os.path.join(rundir, '..', 'python', 'centroid and contour', plot_date_str.replace(' ', ''))
os.makedirs(output_dir, exist_ok=True)

# Load the uploaded horizons_results_delta_au.txt file and process the data
file_path = "horizons_delta_au.txt"

# Read and clean the data
data = []
with open(file_path, "r") as file:
    lines = file.readlines()

for line in lines[1:]:  # Skip header
    columns = line.split()
    try:
        date = datetime.strptime(columns[0], "%Y-%b-%d")  # Format: YYYY-MMM-DD
        delta_au = float(columns[1])  # Extract Delta AU value
        data.append((date, delta_au))
    except ValueError as e:
        print(f"Skipping line due to error: {e}")  # Debugging output
        
# =============================== Initial Coordinates and Background Regions ===============================

if obs_date_str == '20241018':
    x_init_list = [789, 721, 697, 669, 669, 667, 667, 869, 0]
    y_init_list = [653, 645, 641, 637, 641, 639, 639, 585, 0]
    lower_percentage = 80
    contour_level_1 = 3
    contour_level_2 = 6 
    row_number = 1
    exp_cont = 20
    exp_co = 40
    exp_h2o = 20
    rotation = -179.2
    sun_file = "horizons_results_Sun_RA_DEC_181024.txt"
elif obs_date_str == '20241019':
    x_init_list = [661, 637, 613, 591, 590, 591, 569, 543, 621, 595,
                   647, 621, 593, 571, 551, 525, 503, 477, 451, 427, 403]
    y_init_list = [579, 577, 573, 571, 593, 591, 589, 583, 601, 597,
                   555, 555, 555, 551, 547, 545, 543, 539, 535, 545, 533]
    lower_percentage = 75.6
    contour_level_1 = 2
    contour_level_2 = 4
    row_number = 2
    exp_cont = 20
    exp_co = 40
    exp_h2o = 20
    rotation = -179.2
    sun_file = "horizons_results_Sun_RA_DEC_191024.txt"
elif obs_date_str == '20241024':
    x_init_list = [750, 748, 746, 744, 741, 666, 661, 0]
    y_init_list = [514, 518, 517, 517, 520, 512, 511, 0]
    lower_percentage = 60
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 3
    exp_cont = 30
    exp_co = 300
    exp_h2o = 15
    rotation = -179.2
    sun_file = "horizons_results_Sun_RA_DEC_241024.txt"
elif obs_date_str == '20241101':
    x_init_list = [806, 806, 806, 806, 806]
    y_init_list = [603, 603, 603, 603, 603]
    lower_percentage = 34
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 4
    exp_cont = 60
    exp_co = 300
    exp_h2o = 30
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_011124.txt"
elif obs_date_str == '20241102':
    x_init_list = [816, 816, 816, 813, 811, 809, 808]
    y_init_list = [609, 609, 609, 611, 611, 614, 613]
    lower_percentage = 34
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 5
    exp_cont = 60
    exp_co = 300
    exp_h2o = 30
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_021124.txt"
elif obs_date_str == '20241105':
    x_init_list = [636, 633, 633, 629, 626]
    y_init_list = [624, 626, 630, 630, 631]
    lower_percentage = 60
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 6
    exp_cont = 120
    exp_co = 600
    exp_h2o = 60
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_051124.txt"
elif obs_date_str == '20241106':
    x_init_list = [628, 624, 585, 655, 632]
    y_init_list = [591, 592, 597, 600, 603]
    lower_percentage = 70
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 7
    exp_cont = 120
    exp_co = 600
    exp_h2o = 60
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_061124.txt"
elif obs_date_str == '20241109':
    x_init_list = [670, 665, 661, 660, 658]
    y_init_list = [548, 548, 550, 551, 551]
    lower_percentage = 70
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 8
    exp_cont = 120
    exp_co = 600
    exp_h2o = 60
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_091124.txt"
elif obs_date_str == '20241120': 
    x_init_list = [601, 597, 598, 594]
    y_init_list = [601, 603, 606, 607]
    lower_percentage = 70
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 9
    exp_cont = 200
    exp_co = 300
    exp_h2o = 300
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_201124.txt"
elif obs_date_str == '20241121':
    x_init_list = [603, 601, 599, 599]
    y_init_list = [619, 621, 623, 625]
    lower_percentage = 70
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 10
    exp_cont = 200
    exp_co = 300
    exp_h2o = 300
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_211124.txt"
elif obs_date_str == '20241122':
    x_init_list = [599, 596, 592, 591, 588, 586]
    y_init_list = [605, 606, 608, 609, 610, 612]
    lower_percentage = 70
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 11
    exp_cont = 200
    exp_co = 300
    exp_h2o = 300
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_221124.txt"
elif obs_date_str == '20241123':
    x_init_list = [590, 587, 587, 583, 582, 580]
    y_init_list = [592, 594, 594, 598, 598, 600] 
    lower_percentage = 70
    contour_level_1 = 3
    contour_level_2 = 6
    row_number = 12
    exp_cont = 200
    exp_co = 300
    exp_h2o = 300
    rotation = -179.1
    sun_file = "horizons_results_Sun_RA_DEC_231124.txt"
else:
    raise ValueError(f"No configuration found for date {obs_date_str}")
    
#=================Flat Fields Cases===================================    

angle_configs = {
    '20241018': {'Cont': 27.5, 'CO': 27.5, 'H2O': 22.5},
    '20241019': {'Cont': 30, 'CO': 30, 'H2O': 30},
    '20241024': {'Cont684': 22, 'CO': 22, 'H2O': 19},
    '20241101': {'Cont684':20.5, 'CO': 21, 'H2O': 21},
    '20241102': {'Cont684':20, 'CO': 20, 'H2O': 19},
    '20241105': {'Cont684':21, 'CO': 24, 'H2O': 21},
    '20241106': {'Cont684':22, 'CO': 22, 'H2O': 22},
    '20241109': {'Cont684':23, 'CO': 24, 'H2O': 23},
    '20241120': {'Cont':27.5, 'CO': 27.5, 'H2O': 31.5},
    '20241121': {'Cont':27.5, 'CO': 31, 'H2O': 27.5},
    '20241122': {'Cont':29.5, 'CO': 30.5, 'H2O': 28.5},
    '20241123': {'Cont':28.5, 'CO': 31, 'H2O': 29},
}

flat_dirs_by_date = {
    '20241018': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241018','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241018','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241018','', 'H2O')
    },
    '20241019': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241019','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241019','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241019','', 'H2O')
    },
    '20241024': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241024','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241024','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241019','', 'H2O')
    },
    '20241101': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'H2O')
    },
    '20241102': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'H2O')
    },
    '20241105': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'H2O')
    },
    '20241106': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241105','', 'H2O')
    },
    '20241109': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241109','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241109','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241109','', 'H2O')
    },
    '20241120': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241120','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'H2O')
    },
    '20241121': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'H2O')
    },
    '20241122': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241122','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241122','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'H2O')
    },
    '20241123': {
        'Cont': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241122','', 'Cont'),
        'CO': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241123','', 'CO'),
        'H2O': ('/Users/kylerichardaffleck/Downloads/C2023A3/20241121','', 'H2O')
    },
}
# =============================== Utility Functions ===============================

def darken_color(color, amount=0.5):
    """Darken the given color by a certain amount (0 to 1)."""
    try:
        c = mcolors.cnames[color]
    except:
        c = color
    rgb = mcolors.to_rgb(c)
    return [max(0, min(1, channel * amount)) for channel in rgb]
    
def list_files(directory, substring1, substring2):
    return [
        filename for filename in os.listdir(directory)
        if fnmatch.fnmatch(filename, f"{substring1}*{substring2}*")
    ]

def load_science(rundir, runnum, filtname):
    filename = f"{rundir}/C2023A3-{runnum:04d}{filtname}.fit"
    return fits.open(filename)

def create_master_flat(flat_files, obs_time, rundir):
    master_flat = []
    for f in flat_files:
        temp = fits.open(f"{rundir}/{f}")
        flat_time = datetime.strptime(temp[0].header['DATE-OBS'], "%Y-%m-%dT%H:%M:%S.%f")
        if abs(flat_time - obs_time) <= timedelta(days=7):
            flat_data = temp[0].data.astype('float')
            if flat_data.shape != (1024, 1024):
                flat_data = flat_data[:1024, :1024]
            master_flat.append(flat_data / np.median(flat_data))
    if len(master_flat) == 0:
        raise ValueError("No suitable flat frames found within 7 days.")
    return np.median(np.array(master_flat), axis=0)

def refined_centroid(image, x_guess, y_guess, box_size=101):
    image_copy = image.copy()

    # Apply date-specific background removal
    image_copy = subtract_adaptive_background(image_copy)
    half_box = box_size // 2
    x_min = max(0, int(x_guess) - half_box)
    x_max = min(image_copy.shape[1], int(x_guess) + half_box)
    y_min = max(0, int(y_guess) - half_box)
    y_max = min(image_copy.shape[0], int(y_guess) + half_box)

    sub_image = image_copy[y_min:y_max, x_min:x_max]

    if sub_image.shape[0] < 3 or sub_image.shape[1] < 3:
        print("Warning: Subimage too small for centroiding.")
        return None

    if np.isnan(sub_image).any() or np.isinf(sub_image).any():
        print("Warning: Subimage contains NaNs or infs.")
        return None

    try:
        result = centroid_2dg(sub_image)
    except Exception as e:
        print(f"Centroid 2DG error: {e}")
        return None

    if result is None or len(result) != 2:
        print("Centroid result is invalid.")
        return None

    x_centroid_global = x_min + result[0]
    y_centroid_global = y_min + result[1]

    # Sanity check: should be within the image
    if not (0 <= x_centroid_global < image.shape[1] and 0 <= y_centroid_global < image.shape[0]):
        print(f"Invalid centroid location: ({x_centroid_global}, {y_centroid_global})")
        return None

    return x_centroid_global, y_centroid_global

def calculate_physical_size(pixel_numbers, pixel_resolution, delta_au):
    # Convert geocentric distance from AU to km
    D_km = delta_au * 149597870.7  # 1 AU = 149,597,870.7 km

    # Compute total angular span in arcseconds
    Theta_arcsec = pixel_numbers * pixel_resolution

    # Convert to radians
    Theta_rad = Theta_arcsec * (np.pi / (180 * 3600))

    # Compute the exact physical size using trigonometry
    L_km = 2 * D_km * Theta_rad / 2

    return L_km 
    
def extract_radial_profile(image, center, angle_deg, start=-700, end=100):
    angle_rad = np.deg2rad(angle_deg)
    
    x = center[0] + np.cos(angle_rad) * np.arange(start, end)
    y = center[1] + np.sin(angle_rad) * np.arange(start, end)
    profile = map_coordinates(image, [y, x], order=1, mode='nearest')
    return profile, x, y

def load_transmission(file_path):
    data = np.loadtxt(file_path, skiprows=1)  # <-- skip the header row
    wavelength = data[:, 0]
    transmission = data[:, 1]
    return wavelength, transmission

def integrate_transmission(wavelength, transmission):
    return np.trapz(transmission, wavelength)

def align_image_to_reference(image_to_shift, reference_centroid, target_centroid):
    dy = target_centroid[1] - reference_centroid[1]
    dx = target_centroid[0] - reference_centroid[0]
    return shift(image_to_shift, shift=(dy, dx), order=1, mode='nearest')

def get_scaled(image):
    clim1 = np.nanpercentile(np.log10(np.abs(image)), lower_percentage)
    clim2 = np.nanpercentile(np.log10(np.abs(image)), 99.7)
    return np.clip(np.log10(np.abs(image)), clim1, clim2)

def parse_sun_ephemeris(ephemeris_file):
    sun_times = []
    sun_coords = []
    with open(ephemeris_file, 'r') as f:
        lines = f.readlines()
    in_table = False
    for line in lines:
        if "$$SOE" in line:
            in_table = True
            continue
        if "$$EOE" in line:
            break
        if in_table and line.strip():
            parts = line.split()
            try:
                time_str = parts[0] + " " + parts[1]
                obs_time = datetime.strptime(time_str, "%Y-%b-%d %H:%M")
                if parts[2] in ["A", "m", "Am"]:
                    parts.pop(2)
                ra_str = " ".join(parts[2:5])  
                dec_str = " ".join(parts[5:8])  
                sun_coord = SkyCoord(ra_str, dec_str, unit=(u.hourangle, u.deg), frame='icrs')
                sun_times.append(obs_time)
                sun_coords.append(sun_coord)

            except Exception as e:
                print(f"Skipping line: {line.strip()} due to error: {e}")
                continue
    return sun_times, sun_coords

def get_closest_sun_coord(obs_time, sun_times, sun_coords):
    time_deltas = [abs((obs_time - t).total_seconds()) for t in sun_times]
    closest_idx = np.argmin(time_deltas)
    return sun_coords[closest_idx]

def extract_tangential_profile(image, center, angle_deg, offset_along_tail, length=40):
    angle_rad = np.deg2rad(angle_deg)
    tangent_angle = angle_rad + np.pi / 2  # 90° rotation for perpendicular

    x_center = center[0] - np.sin(angle_rad) * offset_along_tail
    y_center = center[1] - np.cos(angle_rad) * offset_along_tail

    t_x = x_center - np.sin(tangent_angle) * np.arange(-length, length + 1)
    t_y = y_center - np.cos(tangent_angle) * np.arange(-length, length + 1)

    profile = map_coordinates(image, [t_y, t_x], order=1, mode='nearest')
    return profile, np.arange(-length, length + 1), t_x, t_y  # returns pixel offsets & actual coords

def subtract_adaptive_background(image, box_size=101):
    sigma_clip = SigmaClip(sigma=3.)
    bkg_estimator = MedianBackground()
    bkg = Background2D(image, (box_size, box_size), filter_size=(3, 3),
                       sigma_clip=sigma_clip, bkg_estimator=bkg_estimator)
    return image - bkg.background

def rayleigh_tau(wavelength_nm, tau_ref=0.1, lambda_ref=550):
    return tau_ref * (lambda_ref / wavelength_nm)**4 
# =============================== Filter Parameters ===============================

# Load transmission profiles
wl_cont, trans_cont = load_transmission("trans_c09_continuo_6840.txt")
wl_co, trans_co = load_transmission("trans_c07_co_4260.txt")
wl_h2o, trans_h2o = load_transmission("trans_c10_h2o_7000.txt")

# Integrate transmission curves
int_cont = integrate_transmission(wl_cont, trans_cont)
int_co = integrate_transmission(wl_co, trans_co)
int_h2o = integrate_transmission(wl_h2o, trans_h2o)

lambda_cont = 684.7  # nm
lambda_co = 431.5    # nm
lambda_h2o = 704.3   # nm

reflectance_cont = 1.148
reflectance_h2o = 1.168
reflectance_co = 0.805

combined_profiles = {
    'Cont': [],
    'CO': [],
    'H2O': []
}

smoothed_profiles = {
    'Cont': [],
    'CO': [],
    'H2O': []
}

shifted_x_values = []  # Store common x-axis
run_labels = [] 

index = 0
init_guess_index = 0

display_titles = {
    'Cont684': r'Dust Continuum',
    'Cont': r'Dust Continuum',
    'CO': r'CO$^+$',
    'H2O': r'H$_2$O$^+$'
}
if obs_date_str in ['20241024','20241101', '20241102', '20241105', '20241106', '20241109']:
    filter_names = ['Cont684', 'CO', 'H2O']
else:
    filter_names = ['Cont', 'CO', 'H2O']
# Define scaling factors per filter
exposure_scalings = {
    'Cont': exp_cont * 3,
    'CO': exp_co,
    'H2O': exp_h2o * 1.5
}
# =============================== Main Loop ===============================

for runnumber in range(1, len(x_init_list) + 1):
    print(f"Processing run number {runnumber} - Using filter names: {filter_names}")
    centroid_dict = {} 
    try:
        x_init, y_init = x_init_list[init_guess_index], y_init_list[init_guess_index]
        sci_raw = [
            load_science(rundir, runnumber, filter_names[0]),
            load_science(rundir, runnumber, filter_names[1]),
            load_science(rundir, runnumber, filter_names[2])
        ]

        header = sci_raw[0][0].header  # Extract header from the first filter image

        # Extract relevant values
        pixel_size_microns = header.get('XPIXSZ', None)  # Pixel size in microns
        focal_length_mm = header.get('FOCALLEN', None)  # Focal length in mm
    
        if pixel_size_microns is not None and focal_length_mm is not None:
            pixel_resolution = 206.265 * pixel_size_microns / focal_length_mm  # Resolution in mm/mm
            print(f"Run {runnumber}: Pixel Resolution = {pixel_resolution:.6f} arcseconds/px")
        else:
            print(f"Run {runnumber}: Missing header values for pixel resolution calculation.")
            
        obs_time = datetime.strptime(sci_raw[0][0].header['DATE-OBS'], "%Y-%m-%dT%H:%M:%S.%f")

        pixel_resolution = 206.265 * pixel_size_microns/focal_length_mm
        obs_time = datetime.strptime(sci_raw[0][0].header['DATE-OBS'], "%Y-%m-%dT%H:%M:%S.%f")

        biases = list_files(rundir, 'Bias', '')
        flat_dirs = flat_dirs_by_date.get(obs_date_str)
        if not flat_dirs:
            raise ValueError(f"No flat field configuration for date {obs_date_str}")
        
        flat_dir_cont = os.path.join(flat_dirs['Cont'][0], flat_dirs['Cont'][1])
        flat_dir_co   = os.path.join(flat_dirs['CO'][0],   flat_dirs['CO'][1])
        flat_dir_h2o  = os.path.join(flat_dirs['H2O'][0],  flat_dirs['H2O'][1])
        
        flats_Cont = list_files(flat_dir_cont, 'Flat_2x2', filter_names[0])
        flats_CO   = list_files(flat_dir_co,   'Flat_2x2', filter_names[1])
        flats_H2O  = list_files(flat_dir_h2o,  'Flat_2x2', filter_names[2])

        master_bias = []
        for f in biases:
            temp = fits.open(f"{rundir}/{f}")
            bias_time = datetime.strptime(temp[0].header['DATE-OBS'], "%Y-%m-%dT%H:%M:%S.%f")
            if abs(bias_time - obs_time) <= timedelta(days=7):
                master_bias.append(temp[0].data.astype('float'))
        if len(master_bias) == 0:
            raise ValueError("No suitable bias frames found within 7 days.")
        master_bias = np.median(np.array(master_bias), axis=0)

        sci_biased = {i: sci_raw[i][0].data.astype('float') - master_bias for i in range(3)}

        master_flat_Cont = create_master_flat(flats_Cont, obs_time, flat_dir_cont)
        master_flat_CO = create_master_flat(flats_CO, obs_time, flat_dir_co)
        master_flat_H2O = create_master_flat(flats_H2O, obs_time, flat_dir_h2o)

        sci_Cont = sci_biased[0] / master_flat_Cont
        sci_CO = sci_biased[1] / master_flat_CO
        sci_H2O = sci_biased[2] / master_flat_H2O
        read_noise_electrons = np.std(master_bias)  # from bias frames
        print(f"Estimated readout noise = {read_noise_electrons:.2f} ADU")
        
        sci_images = [[sci_raw[i][0].data, sci_biased[i], sci_flat]
                      for i, sci_flat in enumerate([sci_Cont, sci_CO, sci_H2O])]
        titles = ['Raw', 'Raw - Bias', '(Raw - Bias)/Flat']

        fig, axes = plt.subplots(3, 3, figsize=(12, 10))

        if 1 <= row_number <= len(data):  
            delta_au_value = data[row_number - 1][1]  # Extract delta_au (2nd column) for given row
            print(f"Using delta_au = {delta_au_value} for row {row_number}")  # Debugging check
        else:
            raise ValueError(f"row_number {row_number} is out of bounds for data with {len(data)} rows.")

        for i, filter_name in enumerate([filter_names[0], filter_names[1], filter_names[2]]):
            for j, (image, title) in enumerate(zip(sci_images[i], titles)):
                ax = axes[i, j]
                ax.tick_params(axis='both', labelsize=16)
                header = sci_raw[i][0].header
                obs_date = datetime.strptime(header['DATE-OBS'], "%Y-%m-%dT%H:%M:%S.%f").strftime("%d-%m-%Y %H:%M:%S")
                
                with np.errstate(divide='ignore', invalid='ignore'):
                    clim1 = np.nanpercentile(np.log10(np.abs(image)), lower_percentage)
                    clim2 = np.nanpercentile(np.log10(np.abs(image)), 99.7)
                    scaled_image = np.clip(np.log10(np.abs(image)), clim1, clim2)
        
                ny, nx = image.shape
                x_pixels = np.arange(nx)
                y_pixels = np.arange(ny)
                x_km = calculate_physical_size(x_pixels, pixel_resolution, delta_au_value)
                y_km = calculate_physical_size(y_pixels, pixel_resolution, delta_au_value)
                extent = [x_km[0], x_km[-1], y_km[0], y_km[-1]]
                ax.set_xticks(np.linspace(extent[0], extent[1], 3)) 
                ax.set_yticks(np.linspace(extent[2], extent[3], 3))
                img = ax.imshow(scaled_image, cmap='Greys', origin='lower', vmin=clim1, vmax=clim2, extent=extent)

                # Parse Sun coordinates
                sun_times, sun_coords = parse_sun_ephemeris(sun_file)
                sun_coord = get_closest_sun_coord(obs_time, sun_times, sun_coords)
                # Parse comet coordinates from header
                comet_ra_str = header['OBJCTRA']  
                comet_dec_str = header['OBJCTDEC']  
                comet_coord = SkyCoord(comet_ra_str, comet_dec_str, unit=(u.hourangle, u.deg), frame='icrs')
        
                if j == 0:
                    #exptime = header.get('EXPTIME', 'N/A')
                    #obj_ra = header.get('OBJCTRA', 'N/A')
                    #obj_dec = header.get('OBJCTDEC', 'N/A')
                    #obj_alt = header.get('OBJCTALT', 'N/A')
                    #obj_az = header.get('OBJCTAZ', 'N/A')
                    #obj_ha = header.get('OBJCTHA', 'N/A')
                    #AM = header.get('AIRMASS', 'N/A')
        
                    if j == 0:
                        display_label = display_titles.get(filter_name, filter_name)
                        ax.set_ylabel(
                            f"{display_label}\n\n",
                        rotation=90, fontsize=16, labelpad=1, va='center')
                
                if j == 2:
                    # Instead of calling refined_centroid twice, do:
                    if filter_name not in centroid_dict:
                        centroid = refined_centroid(scaled_image, x_init, y_init, box_size=101)
                        if centroid is None:
                            print(f"Run {runnumber}: Failed to compute centroid for {filter_name}")
                            continue
                        centroid_dict[filter_name] = centroid
                    x_centroid, y_centroid = centroid_dict[filter_name]
                    # Convert centroid to km
                    x_centroid_km = calculate_physical_size(x_centroid, pixel_resolution, delta_au_value)
                    y_centroid_km = calculate_physical_size(y_centroid, pixel_resolution, delta_au_value)
                    ax.scatter(x_centroid_km, y_centroid_km, color='red', marker='+')

                    #divider = make_axes_locatable(ax)
                    #cax = divider.append_axes("right", size="5%", pad=0.05)
                    #cbar = plt.colorbar(img, cax=cax)
                    #cbar.ax.tick_params(labelsize=16)
                
                    # Retain optional radial profile overlay (optional: comment this out too if needed)
                    angle_deg = angle_configs[obs_date_str][filter_name]
                    profile, x_profile, y_profile = extract_radial_profile(scaled_image, (x_centroid, y_centroid), angle_deg)
                    print(f"Refined Centroid for run {runnumber}, filter {filter_name}: x={x_centroid:.2f}, y={y_centroid:.2f}")
                    # Convert radial profile path to km
                    x_profile_km = calculate_physical_size(x_profile, pixel_resolution, delta_au_value)
                    y_profile_km = calculate_physical_size(y_profile, pixel_resolution, delta_au_value)
                    #ax.plot(x_profile_km, y_profile_km, color='blue', linewidth=0.8, linestyle='--')

                if i == 0:
                    ax.set_title(title, fontsize =16)
        #fig.suptitle(f"Run {runnumber} - {plot_date_str}")
        fig.tight_layout()
        fig.savefig(f"{output_dir}/{plot_date_str.replace(' ', '')}_centroid_contour_{runnumber:04d}.pdf")
        plt.show()

        # ---- NEW FIGURE FOR RADIAL PROFILES ----
        fig_profiles, profile_axes = plt.subplots(1, 3, figsize=(14, 4), sharey=False)
        
        # Create a single figure for combined radial profiles
        fig_combined, ax_combined = plt.subplots(figsize=(7, 5))
        
        colors = ['black', 'red', 'blue']
        distance_pixels = np.arange(-700, 100)  # Original x-axis values
        
        # Ensure row_number is within bounds before extracting delta_au
        if 1 <= row_number <= len(data):  
            delta_au_value = data[row_number - 1][1]  # Extract delta_au (2nd column) for given row
            print(f"Using delta_au = {delta_au_value} for row {row_number}")  # Debugging check
            distance_pixels_corrected = calculate_physical_size(distance_pixels, pixel_resolution, delta_au_value)
        else:
            raise ValueError(f"row_number {row_number} is out of bounds for data with {len(data)} rows.")
        
        for i, (filter_name, color) in enumerate(zip(filter_names, colors)):
            image = sci_images[i][2]  # This is the flat-fielded image (but still raw, not log-scaled)
        
            # Scale image the same way as in the 3x3 plot
            with np.errstate(divide='ignore', invalid='ignore'):
                clim1 = np.nanpercentile(np.log10(np.abs(image)), lower_percentage)
                clim2 = np.nanpercentile(np.log10(np.abs(image)), 100)
                scaled_image = np.clip(np.log10(np.abs(image)), clim1, clim2)
        
            # Find centroid from scaled image
            x_centroid, y_centroid = centroid_dict.get(filter_name, (None, None))
            if x_centroid is None or y_centroid is None:
                print(f"Centroid for {filter_name} is missing, skipping...")
                continue
        
            # Extract profile from scaled image
            angle_deg = angle_configs[obs_date_str][filter_name]
            profile, x_profile, y_profile = extract_radial_profile(scaled_image, (x_centroid, y_centroid), angle_deg)
        
            # Find the index of the maximum peak
            peak_index = np.argmax(profile)
            peak_shift = distance_pixels_corrected[peak_index]  # Get the x value at the peak
        
            # Shift x values so the peak aligns with x = 0
            shifted_x = distance_pixels_corrected - peak_shift

            # Individual subplot for each filter
            ax = profile_axes[i]
            ax.plot(shifted_x, profile, color=color)
            #ax.set_title(f"{filter_name} - Run {runnumber}", fontsize=8)
            ax.set_xlabel("Distance (pixels)", fontsize=7)
            if i == 0:
                ax.set_ylabel("Logged Intensity", fontsize=7)
            ax.grid(True)
            ax.tick_params(axis='both', labelsize=16)
        
            # Plot on the combined figure
            display_label = display_titles.get(filter_name, filter_name)
            ax_combined.plot(shifted_x, profile, color=color, label=display_label)
        
        # Customize combined plot
        #ax_combined.set_title(f"Comparison of Radial Profiles - Run {runnumber}", fontsize=10)
        ax_combined.set_xlabel("Distance (km)", fontsize=16)
        ax_combined.set_ylabel("Logged ADU", fontsize=16)
        ax_combined.legend(fontsize=16)
        ax_combined.grid(True)
        ax_combined.tick_params(axis='both', labelsize=16)
        
        # Save individual and combined plots
        #fig_profiles.suptitle(f"Radial Profiles (Scaled Image) - Run {runnumber}")
        fig_profiles.tight_layout()
        fig_profiles.savefig(f"{output_dir}/{plot_date_str.replace(' ', '')}_radial_profiles_{runnumber:04d}.pdf")

        fig_combined.tight_layout()
        fig_combined.savefig(f"{output_dir}/{plot_date_str.replace(' ', '')}_radial_profiles_combined_{runnumber:04d}.pdf")
       
        # ====== Physically-Informed Continuum Subtraction and Visualization ======

        overlap_min = max(min(wl_cont), min(wl_h2o))
        overlap_max = min(max(wl_cont), max(wl_h2o))
        
        # Interpolate both to same wavelength base over the overlap range
        overlap_wl = np.linspace(overlap_min, overlap_max, 500)
        interp_cont = np.interp(overlap_wl, wl_cont, trans_cont)
        interp_h2o = np.interp(overlap_wl, wl_h2o, trans_h2o)
        
        # Multiply the overlapping parts and integrate
        overlap_integral = np.trapz(interp_cont * interp_h2o, overlap_wl)
        total_cont_integral = integrate_transmission(wl_cont, trans_cont)
        
        beta_h2o = overlap_integral / total_cont_integral
        print(f"Estimated β_H2O = {beta_h2o:.3f}")

        with np.errstate(divide='ignore', invalid='ignore'):
            # Get centroids from flat-fielded, unscaled images
            # SCALE the images BEFORE centroiding
            scaled_Cont = get_scaled(sci_Cont)
            scaled_CO = get_scaled(sci_CO)
            scaled_H2O = get_scaled(sci_H2O)
            
            # Use scaled images for centroiding
            centroid_Cont = centroid_dict.get(filter_names[0])
            centroid_CO = centroid_dict.get(filter_names[1])
            centroid_H2O = centroid_dict.get(filter_names[2])

            if None in (centroid_Cont, centroid_CO, centroid_H2O):
                raise ValueError("One or more centroids could not be computed.")
            
            x_ref, y_ref = centroid_Cont
            x_co, y_co = centroid_CO
            x_h2o, y_h2o = centroid_H2O
            
            # Calculate shifts: how to move CO/H2O to align with Cont
            shift_CO = (y_ref - y_co, x_ref - x_co)   # (dy, dx)
            shift_H2O = (y_ref - y_h2o, x_ref - x_h2o)
            
            print(f"\nRefined Centroids:")
            print(f"Cont:  x={x_ref:.2f}, y={y_ref:.2f}")
            print(f"CO:    x={x_co:.2f}, y={y_co:.2f} → Shift: dx={shift_CO[1]:+.2f}px, dy={shift_CO[0]:+.2f}px")
            print(f"H2O:   x={x_h2o:.2f}, y={y_h2o:.2f} → Shift: dx={shift_H2O[1]:+.2f}px, dy={shift_H2O[0]:+.2f}px")
            
            # Apply shifts to align CO and H2O to Cont
            sci_CO = shift(sci_CO, shift=shift_CO, order=1, mode='nearest')
            sci_H2O = shift(sci_H2O, shift=shift_H2O, order=1, mode='nearest')

        header = sci_raw[0][0].header
        airmass = float(header.get('AIRMASS', 'N/A'))
        print(f"Run {runnumber} — AIRMASS = {airmass:.3f}")
        # Compute τ at each wavelength
        tau_cont = rayleigh_tau(lambda_cont)
        tau_co = rayleigh_tau(lambda_co)
        tau_h2o = rayleigh_tau(lambda_h2o)
        # Normalize all images by their exposure time (important!)
        sci_Cont /= (exp_cont * 3) 
        sci_Cont *= (1- beta_h2o)
        sci_CO /= (exp_co)
        sci_CO *= 15
        sci_H2O /= (exp_h2o * 1.5)
        sigma_Cont = np.sqrt(np.abs(sci_Cont) + read_noise_electrons**2)
        sigma_CO   = np.sqrt(np.abs(sci_CO)   + read_noise_electrons**2)
        sigma_H2O  = np.sqrt(np.abs(sci_H2O)  + read_noise_electrons**2)
        # Recalculate alpha coefficients WITHOUT exposure time ratio now
        alpha_co_phys = (reflectance_co / reflectance_cont) * (1 - beta_h2o)  
        alpha_h2o_phys = (reflectance_h2o / reflectance_cont) * (1 - beta_h2o) 
        
        print(f"Normalized & Scaled α_CO = {alpha_co_phys:.4f}, α_H2O = {alpha_h2o_phys:.4f}")
        
        # Now that sci_CO and sci_H2O are aligned to sci_Cont and normalized, we can subtract directly
        co_clean = sci_CO * (int_cont/int_co) - (alpha_co_phys * sci_Cont)
        h2o_clean = sci_H2O * (int_cont/int_h2o) - (alpha_h2o_phys * sci_Cont)
        # Apply Gaussian smoothing to the entire cleaned images
        sigma = 3  
        co_clean_smoothed = gaussian_filter(co_clean, sigma=sigma)
        h2o_clean_smoothed = gaussian_filter(h2o_clean, sigma=sigma)
        cont_smoothed = gaussian_filter(sci_Cont, sigma=sigma)
        co_clean_smoothed /= 15
        sci_images_cleaned_simple = [
            [co_clean_smoothed, 'CO - Cleaned'],
            [h2o_clean_smoothed, 'H2O - Cleaned']
        ]
        fig_clean_dual, axes = plt.subplots(2, 2, figsize=(12, 10))  # 2 rows (filters), 2 cols (annotated vs clean)

        for i, (img_data, title_label) in enumerate(sci_images_cleaned_simple):
            # Clean-only panel
            ny, nx = img_data.shape
            x_pixels = np.arange(nx)
            y_pixels = np.arange(ny)
            x_km = calculate_physical_size(x_pixels, pixel_resolution, delta_au_value)
            y_km = calculate_physical_size(y_pixels, pixel_resolution, delta_au_value)
            extent = [x_km[0], x_km[-1], y_km[0], y_km[-1]]

            # Create subplot axes
            ax_annotated = axes[i, 0]
            ax_clean = axes[i, 1]

            with np.errstate(divide='ignore', invalid='ignore'):
                clim1 = np.nanpercentile(np.log10(np.abs(img_data)), lower_percentage)
                clim2 = np.nanpercentile(np.log10(np.abs(img_data)), 99.7)
                scaled = np.clip(np.log10(np.abs(img_data)), clim1, clim2)

            # Plot annotated
            im1 = ax_annotated.imshow(scaled, origin='lower', cmap='Greys', vmin=clim1, vmax=clim2, extent=extent)
            filter_key = filter_names[i+1]
            centroid = centroid_dict.get(filter_key)
            if centroid is not None:
                x_px, y_px = centroid
                x_km_centroid = calculate_physical_size(x_px, pixel_resolution, delta_au_value)
                y_km_centroid = calculate_physical_size(y_px, pixel_resolution, delta_au_value)
                ax_annotated.set_xticks(np.linspace(extent[0], extent[1], 3)) 
                ax_annotated.set_yticks(np.linspace(extent[2], extent[3], 3))
                ax_annotated.scatter(x_km_centroid, y_km_centroid, color='red', marker='+')
                ax_annotated.tick_params(axis='both', labelsize=16)
                angle = angle_configs[obs_date_str][filter_key]
                profile, x_p, y_p = extract_radial_profile(scaled, (x_px, y_px), angle)
                x_km_profile = calculate_physical_size(x_p, pixel_resolution, delta_au_value)
                y_km_profile = calculate_physical_size(y_p, pixel_resolution, delta_au_value)
                ax_annotated.plot(x_km_profile, y_km_profile, color='blue', linestyle='--', linewidth=0.8, alpha=0.5)
                sky_pa = comet_coord.position_angle(sun_coord).to(u.deg).value
                anti_solar_pa = (sky_pa + 180) % 360
                anti_solar_image_angle = (anti_solar_pa - rotation) % 360
                print(anti_solar_image_angle)
                tail_start = -700
                offsets = {
                #'Centroid': 0,
                'Three Eigths': 2 * tail_start / 8,
                'Halfway': tail_start / 2,
                'Six Eights': (5 * tail_start) / 8
            }

                for label, offset in offsets.items():
                    angle_rad = np.deg2rad(anti_solar_image_angle)
                    tangent_angle = angle_rad + np.pi / 2
                    
                    x_base = x_px - np.sin(angle_rad) * offset
                    y_base = y_px - np.cos(angle_rad) * offset
                    
                    # Convert base point to km
                    x_base_km = calculate_physical_size(x_base, pixel_resolution, delta_au_value)
                    y_base_km = calculate_physical_size(y_base, pixel_resolution, delta_au_value)
                    
                    # Tangential extent (25 pixels each side, total 51)
                    dx = 40 * np.sin(tangent_angle)
                    dy = 40 * np.cos(tangent_angle)
                    
                    dx_km = calculate_physical_size(x_base + dx, pixel_resolution, delta_au_value) - x_base_km
                    dy_km = calculate_physical_size(y_base + dy, pixel_resolution, delta_au_value) - y_base_km
                    
                    # Draw tangential sampling line on annotated image
                    ax_annotated.plot(
                        [x_base_km - dx_km, x_base_km + dx_km],
                        [y_base_km - dy_km, y_base_km + dy_km],
                        color='red', linestyle='-', linewidth=1.5, alpha=0.8
                    )
                    

                sky_pa = comet_coord.position_angle(sun_coord).to(u.deg).value
                anti_solar_pa = (sky_pa + 180) % 360
                anti_solar_image_angle = (anti_solar_pa - rotation) % 360
                dx_arrow = 200 * np.sin(np.deg2rad(anti_solar_image_angle))
                dy_arrow = 200 * np.cos(np.deg2rad(anti_solar_image_angle))
                dx_arrow_km = calculate_physical_size(x_px + dx_arrow, pixel_resolution, delta_au_value) - x_km_centroid
                dy_arrow_km = calculate_physical_size(y_px + dy_arrow, pixel_resolution, delta_au_value) - y_km_centroid
                print(anti_solar_image_angle)
                ax_annotated.arrow(
                    x_km_centroid, y_km_centroid,
                    dx_arrow_km, dy_arrow_km,
                    head_width=10, head_length=15, fc='orange', ec='orange', linewidth=1.5
                )
            ax_annotated.set_xlabel("X Distance (km)", fontsize = 16)
            ax_annotated.set_ylabel("Y Distance (km)", fontsize = 16)
            # Clean-only panel
            im2 = ax_clean.imshow(scaled, origin='lower', cmap='Greys', vmin=clim1, vmax=clim2, extent=extent)
            if "CO" in title_label:
                annotated_title = r"CO$^+$ – Cleaned and Annotated"
                clean_title = r"CO$^+$ – Cleaned"
            elif "H2O" in title_label:
                annotated_title = r"H$_2$O$^+$ – Cleaned and Annotated"
                clean_title = r"H$_2$O$^+$ – Cleaned"
            else:
                annotated_title = title_label + " – Cleaned and Annotated"
                clean_title = title_label + " – Cleaned"
            
            ax_annotated.set_title(annotated_title, fontsize=16)
            ax_clean.set_title(clean_title, fontsize=16)
            ax_clean.set_xlabel("X Distance (km)", fontsize = 16)
            ax_clean.set_ylabel("Y Distance (km)", fontsize = 16)
            ax_clean.set_xticks(np.linspace(extent[0], extent[1], 3)) 
            ax_clean.set_yticks(np.linspace(extent[2], extent[3], 3))
            ax_clean.tick_params(axis='both', labelsize=16)
            #cbar_clean = ax_clean.figure.colorbar(im2, ax=ax_clean, shrink=0.7, pad=0.02)
            #cbar_clean.ax.tick_params(labelsize=16)
            divider_clean = make_axes_locatable(ax_clean)
            cax_clean = divider_clean.append_axes("right", size="5%", pad=0.05)
            cbar_clean = plt.colorbar(im2, cax=cax_clean)
            cbar_clean.ax.tick_params(labelsize=16)

        #fig_clean_dual.suptitle(f"Cleaned Ion Images (Annotated vs Clean) - Run {runnumber}", fontsize=13)
        fig_clean_dual.tight_layout()
        fig_clean_dual.savefig(f"{output_dir}/{plot_date_str.replace(' ', '')}_ion_annotated_vs_clean_{runnumber:04d}.pdf")
        plt.show()

        # -------- Radial Profiles After Continuum Subtraction --------
        fig_phys_profiles, ax_phys = plt.subplots(figsize=(7, 5))
        colors_dict = {'Cont': 'black', 'CO': 'red', 'H2O': 'blue'}
        labels_dict = {
            key: display_titles.get(key, key) for key in ['Cont', 'CO', 'H2O']
        }
        
        distance_pixels = np.arange(-700, 100)
        delta_au_value = data[row_number - 1][1]
        distance_pixels_corrected = calculate_physical_size(distance_pixels, pixel_resolution, delta_au_value)
        
        shifted_x_this_run = None  # Store this once per run
        # Define scaling factors per filter

        for idx, (img_data, filt_key) in enumerate(zip([cont_smoothed, co_clean_smoothed, h2o_clean_smoothed], ['Cont', 'CO', 'H2O'])):
            with np.errstate(divide='ignore', invalid='ignore'):
                scaled = np.log10(np.abs(img_data))
        
            # Correct key for centroid lookup
            centroid = centroid_dict.get(filter_names[idx])
            if centroid is None:
                print(f"Skipping centroid plot for {filter_names[idx]} in cleaned image due to missing centroid.")
                continue
            x_centroid, y_centroid = centroid
        
            # Use correct angle config
            angle = angle_configs[obs_date_str][filter_names[idx]]
            profile, _, _ = extract_radial_profile(scaled, (x_centroid, y_centroid), angle)
            # Extract profile errors
            
            # Store for evolution plots
            peak_index = np.argmax(profile)
            peak_shift = distance_pixels_corrected[peak_index]
            shifted_x = distance_pixels_corrected - peak_shift
            combined_profiles[filt_key].append(profile)
            
            if shifted_x_this_run is None:
                shifted_x_this_run = shifted_x
            window_size = 3  # You can adjust (3 = ±1 pixel around)
            half_window = window_size // 2
            errors = np.zeros_like(profile)
            for i in range(len(profile)):
                i_min = max(0, i - half_window)
                i_max = min(len(profile), i + half_window + 1)
                errors[i] = np.nanstd(profile[i_min:i_max])
            # Darken the color
            if filt_key == 'Cont':
                sigma_image = sigma_Cont
            elif filt_key == 'CO':
                sigma_image = sigma_CO
            elif filt_key == 'H2O':
                sigma_image = sigma_H2O
            exposure_scaling = exposure_scalings[filt_key]
            # Compute profile errors from original (linear) image
            profile_errors = map_coordinates(sigma_image, [y_profile, x_profile], order=1, mode='nearest')
            
            # Compute the profile (already log-scaled) and get the linear values again
            linear_profile = 10 ** profile  # Convert back to linear units
            
            # Avoid divide-by-zero
            linear_profile[linear_profile <= 0] = np.nan
            
            # Apply Gaussian error propagation in log space
            dark_color = darken_color(colors_dict[filt_key], amount=0.6)
            
            linear_profile = 10 ** profile  # Convert profile from log10 back to linear units
            errors_corrected = profile_errors / (linear_profile * np.log(10)) 
            errors_corrected *= np.sqrt(effective_pixels)
            errors_corrected *= errors * np.log10 (exposure_scaling)
            # Thin the points
            step = 20
            thinned_x = shifted_x[::step]
            thinned_profile = profile[::step]
            thinned_errors = errors_corrected[::step]
                        
            ax_phys.plot(shifted_x, profile, color=colors_dict[filt_key], label=labels_dict[filt_key])

            # Then overlay error bars (as thin errorbars only, no new line)
            ax_phys.errorbar(thinned_x, thinned_profile, yerr=thinned_errors, fmt='none', ecolor=dark_color, capsize=3, elinewidth=1, alpha=0.8)
            ax_phys.tick_params(axis='both', labelsize=16)
            if filt_key == 'Cont':
                dust_profile = profile
            elif filt_key == 'H2O':
                h2o_profile = profile

            sky_pa = comet_coord.position_angle(sun_coord).to(u.deg).value
            anti_solar_pa = (sky_pa + 180) % 360
            anti_solar_image_angle = (anti_solar_pa - rotation) % 360
            print(anti_solar_image_angle)
            # Tangential profile plotting inside the loop for cleaned radial profiles
            
            tangential_profiles = []
            
            # Define tangential offsets relative to the radial direction
            tail_start = -700
            offsets = {
                #'Centroid': 0,
                'Three Eigths': 2 * tail_start / 8,
                'Halfway':  (9 * tail_start) / 16,
                'Six Eights': (5 * tail_start) / 8
            }
            
            tangential_profiles = []

            for label, offset in offsets.items():
                profile, tang_x_vals, _, _ = extract_tangential_profile(
                    scaled, (x_centroid, y_centroid), anti_solar_image_angle, offset, length=40
                )
            
                tang_x_vals_km = calculate_physical_size(x_centroid + tang_x_vals, pixel_resolution, delta_au_value) - calculate_physical_size(x_centroid, pixel_resolution, delta_au_value)
            
                offset_base_px = x_centroid - np.sin(np.deg2rad(anti_solar_image_angle)) * offset
                offset_km = calculate_physical_size(offset_base_px, pixel_resolution, delta_au_value) - calculate_physical_size(x_centroid, pixel_resolution, delta_au_value)
            
                label_km = f"{-1*offset_km:.0f} km"
                tangential_profiles.append((label_km, tang_x_vals_km, profile))
            
            # Plotting
            fig_tang, ax_tang = plt.subplots(figsize=(7, 4))
            window_size = 3  # choose how many pixels to include around each tangential point for error
            half_window = window_size // 2
            
            for label_km, x_vals_km, profile in tangential_profiles:
                # Compute local standard deviation around each point
               # Get the original image and readout noise
                if filt_key == 'Cont':
                    sigma_image = sigma_Cont
                elif filt_key == 'CO':
                    sigma_image = sigma_CO
                elif filt_key == 'H2O':
                    sigma_image = sigma_H2O
                effective_pixels = 6
                # Map sigma to tangential coordinates
                angle_rad = np.deg2rad(anti_solar_image_angle)
                tangent_angle = angle_rad + np.pi / 2
                x_center = x_centroid - np.sin(angle_rad) * offset
                y_center = y_centroid - np.cos(angle_rad) * offset
                
                t_x = x_center - np.sin(tangent_angle) * np.arange(-40, 41)
                t_y = y_center - np.cos(tangent_angle) * np.arange(-40, 41)
                # Extract sigma values at tangential locations
                sigma_values = map_coordinates(sigma_image, [t_y, t_x], order=1, mode='nearest')
                
                # Convert profile from log10 back to linear units
                linear_profile = 10 ** profile
                linear_profile[linear_profile <= 0] = np.nan 
                # Smooth the log-profile to estimate local variation (Gaussian-style)
                window_size = 3  # or use same smoothing window as radial
                half_window = window_size // 2
                errors = np.zeros_like(profile)
                for k in range(len(profile)):
                    i_min = max(0, k - half_window)
                    i_max = min(len(profile), k + half_window + 1)
                    errors[k] = np.nanstd(profile[i_min:i_max])
                
                # Apply full error propagation logic (same as radial)
                profile_error_log = sigma_values / (linear_profile * np.log(10))
                profile_error_log *= np.sqrt(effective_pixels)
                profile_error_log *= errors * np.log10(exposure_scaling)
                
                peaks, _ = find_peaks(profile)
                # Plot the main full tangential profile first

                f# Plot the main full tangential profile first
                ax_tang.plot(x_vals_km, profile, label=label_km)
                
                # Find all peaks
                peaks, _ = find_peaks(profile)
                
                # If there are peaks, mark only the HIGHEST one
                if len(peaks) > 0:
                    highest_peak_idx = peaks[np.argmax(profile[peaks])]
                    peak_position_km = x_vals_km[highest_peak_idx]
                    print(f"Run {runnumber}, Filter {filter_names[idx]}, {label_km}: Peak at {peak_position_km:.1f} km")
                    ax_tang.plot(x_vals_km[highest_peak_idx], profile[highest_peak_idx], marker='x', color='black', markersize=10)
                
                # Thin the points for errorbars (optional, improves readability)
                step = 10  # or whatever number you like (fewer errorbars)
                thinned_x = x_vals_km[::step]
                thinned_profile = profile[::step]
                thinned_errors = profile_error_log[::step]
                
                # Now overlay ONLY error bars
                ax_tang.errorbar(
                    thinned_x, thinned_profile, yerr=thinned_errors,
                    fmt='none', ecolor='gray', capsize=3, elinewidth=0.8, alpha=0.8
                )

            
            #ax_tang.set_title(f"Tangential Intensity Profiles - Run {runnumber} ({filter_names[idx]})", fontsize=10)
            ax_tang.set_xlabel("Tangential Offset (km)", fontsize=16)
            ax_tang.set_ylabel("Logged ADU", fontsize=16)
            ax_tang.tick_params(axis='both', labelsize =16)
            ax_tang.grid(True)
            #ax_tang.legend(fontsize=16, loc='upper right')
            fig_tang.tight_layout()
            fig_tang.savefig(
                f"{output_dir}/{plot_date_str.replace(' ', '')}_tangential_profile_{filter_names[idx]}_{runnumber:04d}.pdf"
            )

            
        # Now add this part once all filters are done
        shifted_x_values.append(shifted_x_this_run)
        run_labels.append(f"Run {runnumber}")
        
        #ax_phys.set_title(f"Radial Profiles After Continuum Subtraction - Run {runnumber}")
        ax_phys.set_xlabel("Distance (km)", fontsize = 16)
        ax_phys.set_ylabel("Logged ADU", fontsize = 16)
        ax_phys.grid(True)
        ax_phys.legend(fontsize = 16)
        fig_phys_profiles.tight_layout()
        fig_phys_profiles.savefig(f"{output_dir}/{plot_date_str.replace(' ', '')}_radial_profiles_cleaned_{runnumber:04d}.pdf")
        plt.show()

        if 'h2o_profile' in locals() and 'dust_profile' in locals():
            # Convert log-scale profiles to linear intensity values
            dust_peak_linear = 10 ** np.nanmax(dust_profile)
            h2o_peak_linear = 10 ** np.nanmax(h2o_profile)
            if h2o_peak_linear != 0:
                true_peak_ratio = (dust_peak_linear) / (h2o_peak_linear)
                print(f"True Dust / H₂O Ratio (Run {runnumber}): {true_peak_ratio:.3f}")
            else:
                print(f"True Dust / H₂O Ratio (Run {runnumber}): Undefined (H₂O peak is zero)")
        
        # If this is the last run, generate the evolution plots
        if runnumber == len(x_init_list):
            filters = ['Cont', 'CO', 'H2O']
            colors = plt.cm.viridis(np.linspace(0, 1, len(run_labels)))  # Color by run
            distance_pixels = np.arange(-700, 100)  # same as in main loop
            delta_au_value = data[row_number - 1][1]
            distance_pixels_corrected = calculate_physical_size(distance_pixels, pixel_resolution, delta_au_value)
            
            for filt in filters:
                plt.figure(figsize=(8, 5))
                for i, (profile, label) in enumerate(zip(combined_profiles[filt], run_labels)):
                    # Shift x-axis to center on peak
                    peak_index = np.argmax(profile)
                    peak_shift = distance_pixels_corrected[peak_index]
                    shifted_x = distance_pixels_corrected - peak_shift
            
                    plt.plot(shifted_x, profile, label=label, color=colors[i])
            
                plt.title(f"Evolution of Cleaned Radial Profiles - {filt}")
                plt.xlabel("Distance from Peak (km)")
                plt.ylabel("Logged Intensity")
                plt.grid(True)
                plt.legend(fontsize=6)
                plt.tight_layout()
                plt.savefig(f"{output_dir}/{plot_date_str.replace(' ', '')}_evolution_profiles_{filt}.pdf")
                plt.show()
                
        init_guess_index += 1 
        index += 1
    except FileNotFoundError as e:
        print(f"File not found for run number {runnumber}: {e}")
        continue
    except ValueError as e:
        print(f"Skipping run number {runnumber} due to error: {e}")
        continue
```
